```python
from IPython.display import display, Markdown
import seaborn as sns
df = sns.load_dataset('titanic')
```

```python
def section(title, content):
    print( title)
    display(content)
```

# Regla para duplicate:
Duplicado en columna de id es error casi seguro.
Duplicados en todas las columnas con id presente, registro procesado dos veces, se borra.
Duplicado en todas las columnas SIN id: sospechoso, hay que mirarlo. Puede ser coincidencia.

# Bloque de exploracion:
Reportar duplicados, no eliminarlos.

# Bloque de limpieza:
Decision de borrar. Caso por caso


```python
def explore(df, target=None):
    dim = df.shape
    print("dimensions:", dim)
    h = df.head()
    section("Head", h)
    des = df.describe()
    section("Describe:", des)
    ty = df.dtypes
    section("Types:", ty)
    dup = df.duplicated().sum()
    section("Duplicated:", dup)  
```

```python
explore(df)
```
